```python
import pandas as pd
import numpy as np

df = pd.read_csv("indian_startups_funding.csv")

# Rename columns
df.columns = [
    'startup_name',
    'city',
    'industry',
    'funding_stage',
    'amount',
    'funding_date',
    'investors',
    'founded_year'
]

# Clean startup name and city
df['startup_name'] = df['startup_name'].str.strip().str.title()
df['city'] = df['city'].str.strip().str.title()

# Standardize cities
city_map = {
    'Bangalore': 'Bengaluru',
    'Blr': 'Bengaluru',
    'Bombay': 'Mumbai',
    'Gurgaon': 'Gurugram',
    'Ggn': 'Gurugram',
    'Delhi': 'New Delhi',
    'Calcutta': 'Kolkata',
    'Madras': 'Chennai',
    'Hyd': 'Hyderabad',
    'Bom': 'Mumbai'
}

df['city'] = df['city'].replace(city_map)

# Standardize industries
industry_map = {
    'ecommerce': 'E-commerce',
    'e-commerce': 'E-commerce',
    'beautytech': 'BeautyTech',
    'beauty-tech': 'BeautyTech',
    'agritech': 'AgriTech',
    'agri tech': 'AgriTech',
    'agri-tech': 'AgriTech',
    'edtech': 'Edtech',
    'ed-tech': 'Edtech',
    'food tech': 'FoodTech',
    'food-tech': 'FoodTech',
    'foodtech': 'FoodTech',
    'saas': 'SaaS',
    'travel-tech': 'TravelTech',
    'traveltech': 'TravelTech',
    'ai ml': 'AI/ML',
    'aiml': 'AI/ML',
    'ai-ml': 'AI/ML',
    'ai/ml': 'AI/ML',
    'health-tech': 'HealthTech',
    'health tech': 'HealthTech',
    'healthtech': 'HealthTech',
    'electric-vehicles': 'EV',
    'electric vehicles': 'EV',
    'ev': 'EV',
    'fin-tech': 'Fintech',
    'fintech': 'Fintech',
    'ft': 'Fintech',
    'proptech': 'PropTech',
    'prop-tech': 'PropTech',
    'logistics': 'Logistics',
    'gaming': 'Gaming',
    'games': 'Gaming'
}

df['industry'] = (
    df['industry']
    .str.strip()
    .str.lower()
    .replace(industry_map)
)

# Standardize funding stage
stage_map = {
    'series-e': 'Series E',
    'series e': 'Series E',
    'series-d': 'Series D',
    'series d': 'Series D',
    'series-c': 'Series C',
    'series c': 'Series C',
    'series-b': 'Series B',
    'series b': 'Series B',
    'series_a': 'Series A',
    'series a': 'Series A',
    'series-a': 'Series A',
    'pre-a': 'Pre-Series A',
    'pre-series a': 'Pre-Series A',
    'pre series a': 'Pre-Series A',
    'pre-seed': 'Pre-Seed',
    'pre seed': 'Pre-Seed',
    'preseed': 'Pre-Seed',
    'pre-ipo': 'Pre-IPO',
    'seed': 'Seed',
    'bridge': 'Bridge',
    'ipo': 'IPO'
}

df['funding_stage'] = (
    df['funding_stage']
    .str.strip()
    .str.lower()
    .replace(stage_map)
)

# Convert amount to numeric
df['amount'] = (
    df['amount']
    .astype('string')
    .str.replace('₹', '', regex=False)
    .str.replace('INR', '', regex=False)
    .str.replace('Crore', '', regex=False)
    .str.replace('Cr', '', regex=False)
    .str.replace(',', '', regex=False)
    .str.strip()
    .replace('Undisclosed', np.nan)
)

df['amount'] = pd.to_numeric(df['amount'], errors='coerce')

# Convert dates
df['funding_date'] = pd.to_datetime(
    df['funding_date'],
    errors='coerce',
    dayfirst=True
)

# Drop duplicates
df = df.drop_duplicates()
```

```python
# ======Q30======

def get_tier(amount):
    if pd.isna(amount):
        return 'Small / Unknown'
    elif amount >= 1000:
        return 'Unicorn'
    elif amount >= 500:
        return 'Mega'
    elif amount >= 100:
        return 'Large'
    elif amount >= 10:
        return 'Mid'
    else:
        return 'Small / Unknown'

df['funding_tier'] = df['amount'].apply(get_tier)

print(df['funding_tier'].value_counts())
```

```python
# =======Q31=======

bins = [-float('inf'), 2009, 2017, float('inf')]

df['era'] = pd.cut(
    df['founded_year'],
    bins=bins,
    labels=['Early', 'Growth', 'Recent']
)

print(df['era'].value_counts())
```
